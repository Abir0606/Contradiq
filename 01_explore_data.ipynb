```python
from pathlib import Path

import pandas as pd
```

```python
df = pd.read_csv("D:/Contradiq/data/raw/CUAD_v1/master_clauses.csv")
df.head(3)
```

```python
from pypdf import PdfReader

# grab the first contract's path
sample_row = df.iloc[0]
filename = sample_row["Filename"]
print("Filename:", filename)

# find it (it's nested in a subfolder)
contract_root = Path("data/raw/CUAD_v1/full_contract_pdf")
matches = list(contract_root.rglob(filename))
print("Found at:", matches)

reader = PdfReader(matches[0])
print(f"\nNumber of pages: {len(reader.pages)}")
print("\n--- First page text ---\n")
print(reader.pages[0].extract_text()[:1500])
```
